```python
import os, random
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

IMG_SIZE = 96
SEED = 42
random.seed(SEED)

pv_color = "/kaggle/input/datasets/abdallahalidev/plantvillage-dataset/color"
pd_train = "/kaggle/input/datasets/nirmalsankalana/plantdoc-dataset/train"
pd_test  = "/kaggle/input/datasets/nirmalsankalana/plantdoc-dataset/test"

# mapping: pv folder name -> pd folder name (Target_Spot and Spider_mites dropped)
class_map = {
    "Tomato___Bacterial_spot": "Tomato_leaf_bacterial_spot",
    "Tomato___Early_blight": "Tomato_Early_blight_leaf",
    "Tomato___Late_blight": "Tomato_leaf_late_blight",
    "Tomato___Leaf_Mold": "Tomato_mold_leaf",
    "Tomato___Septoria_leaf_spot": "Tomato_Septoria_leaf_spot",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus": "Tomato_leaf_yellow_virus",
    "Tomato___Tomato_mosaic_virus": "Tomato_leaf_mosaic_virus",
    "Tomato___healthy": "Tomato_leaf",
}
pv_classes = list(class_map.keys())
label_of = {c: i for i, c in enumerate(pv_classes)}

def load_images(folder, limit=None):
    files = sorted(os.listdir(folder))
    if limit:
        files = files[:limit]
    imgs = []
    for f in files:
        try:
            img = tf.keras.utils.load_img(os.path.join(folder, f), target_size=(IMG_SIZE, IMG_SIZE))
            imgs.append(tf.keras.utils.img_to_array(img))
        except Exception:
            continue
    return imgs

# --- Build PlantVillage train/val set (capped per class for speed) ---
CAP = 400
X, y = [], []
for pv_c in pv_classes:
    folder = os.path.join(pv_color, pv_c)
    imgs = load_images(folder, limit=CAP)
    X.extend(imgs)
    y.extend([label_of[pv_c]] * len(imgs))

X = np.array(X, dtype="float32") / 255.0
y = np.array(y)
print("PlantVillage set:", X.shape, y.shape)

# shuffle + split 85/15
idx = np.arange(len(X))
np.random.seed(SEED)
np.random.shuffle(idx)
X, y = X[idx], y[idx]
split = int(0.85 * len(X))
X_train, y_train = X[:split], y[:split]
X_val, y_val = X[split:], y[split:]

# --- Build PlantDoc eval set (train+test combined, since both are small) ---
Xp, yp = [], []
for pv_c, pd_c in class_map.items():
    for src in (pd_train, pd_test):
        folder = os.path.join(src, pd_c)
        if os.path.isdir(folder):
            imgs = load_images(folder)
            Xp.extend(imgs)
            yp.extend([label_of[pv_c]] * len(imgs))

Xp = np.array(Xp, dtype="float32") / 255.0
yp = np.array(yp)
print("PlantDoc eval set:", Xp.shape, yp.shape)

# --- Lightweight CNN ---
model = models.Sequential([
    layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)),
    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(len(pv_classes), activation="softmax"),
])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])

model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=5, batch_size=32, verbose=2)

# --- Evaluate on both ---
pv_loss, pv_acc = model.evaluate(X_val, y_val, verbose=0)
pd_loss, pd_acc = model.evaluate(Xp, yp, verbose=0)

print(f"\nPlantVillage val accuracy: {pv_acc:.3f}")
print(f"PlantDoc (real-world) accuracy: {pd_acc:.3f}")
print(f"Accuracy drop: {(pv_acc - pd_acc):.3f}")
```

```python
import tensorflow as tf
import os
import pandas as pd

print("GPU available:", tf.config.list_physical_devices('GPU'))

pv_color = "/kaggle/input/datasets/abdallahalidev/plantvillage-dataset/color"

class_counts = {}
for cls in sorted(os.listdir(pv_color)):
    cls_path = os.path.join(pv_color, cls)
    if os.path.isdir(cls_path):
        class_counts[cls] = len(os.listdir(cls_path))

df_counts = pd.DataFrame(list(class_counts.items()), columns=["class", "count"])
df_counts = df_counts.sort_values("count", ascending=False).reset_index(drop=True)

print(f"Total classes: {len(df_counts)}")
print(f"Total images: {df_counts['count'].sum()}")
print(f"\nLargest class: {df_counts.iloc[0]['class']} ({df_counts.iloc[0]['count']})")
print(f"Smallest class: {df_counts.iloc[-1]['class']} ({df_counts.iloc[-1]['count']})")
print(f"Imbalance ratio (largest/smallest): {df_counts.iloc[0]['count'] / df_counts.iloc[-1]['count']:.1f}x")

print("\nFull distribution:")
print(df_counts.to_string(index=False))
```

```python
''' Step 2: Build the data pipeline

Using `image_dataset_from_directory` for an 85/15 train/validation split, with data
augmentation applied only to the training set (never to validation — augmenting validation
data would let the model "practice" on artificially varied versions of images it's being
graded on, giving a falsely optimistic score). Class weights are computed to counteract the
36x imbalance found above.'''
```

```python
import numpy as np

IMG_SIZE = 224   # MobileNetV2's native input size
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    pv_color, validation_split=0.15, subset="training", seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE), batch_size=BATCH_SIZE
)
val_ds = tf.keras.utils.image_dataset_from_directory(
    pv_color, validation_split=0.15, subset="validation", seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE), batch_size=BATCH_SIZE
)

class_names = train_ds.class_names
print(f"Classes: {len(class_names)}")

# augmentation — training only
augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.08),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomContrast(0.1),
])

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.map(lambda x, y: (augmentation(x, training=True), y), num_parallel_calls=AUTOTUNE)
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

# class weights: total_samples / (num_classes * class_count)
counts = df_counts.set_index("class").loc[class_names, "count"].values
total = counts.sum()
class_weights = {i: total / (len(class_names) * c) for i, c in enumerate(counts)}

print("\nSample class weights (highest = rarest classes):")
sorted_weights = sorted(class_weights.items(), key=lambda x: -x[1])[:5]
for idx, w in sorted_weights:
    print(f"  {class_names[idx]}: {w:.2f}")
```

```python
''' Step 3: Build the model — MobileNetV2 transfer learning

Two-phase training strategy:
1. **Phase 1 (feature extraction):** freeze MobileNetV2's pretrained layers entirely, only
   train a new classification head on top. This adapts the model quickly without risking
   damage to the general-purpose features it already learned from ImageNet.
2. **Phase 2 (fine-tuning):** unfreeze the top portion of MobileNetV2 and continue training
   at a much lower learning rate, letting it adjust its higher-level features specifically
   for leaf textures and lesions. We'll only do this once Phase 1 is stable.

This section builds and runs Phase 1.'''
```

```python
from tensorflow.keras import layers, models, optimizers, callbacks

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3), include_top=False, weights="imagenet"
)
base_model.trainable = False  # freeze for Phase 1

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = tf.keras.applications.mobilenet_v2.preprocess_input(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(len(class_names), activation="softmax")(x)
model = models.Model(inputs, outputs)

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

cb = [
    callbacks.EarlyStopping(monitor="val_accuracy", patience=4, restore_best_weights=True),
    callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-6),
    callbacks.ModelCheckpoint("phase1_best.keras", monitor="val_accuracy", save_best_only=True),
]

history1 = model.fit(
    train_ds, validation_data=val_ds,
    epochs=15, class_weight=class_weights, callbacks=cb
)
```

```python

# model = tf.keras.models.load_model("phase1_best.keras")
# base_model = model.get_layer("mobilenetv2_1.00_224")  # name may vary — check model.summary()

base_model.trainable = True

# MobileNetV2 has ~154 layers. Unfreezing everything risks destroying the
# general ImageNet features Phase 1 relied on — instead, only unfreeze the
# later layers, which hold more task-specific features. Layers before this
# index stay frozen exactly as in Phase 1.
FINE_TUNE_AT = 100

for layer in base_model.layers[:FINE_TUNE_AT]:
    layer.trainable = False

# BatchNorm layers are a special case: even in the "unfrozen" region, keep them
# frozen. Their running mean/variance were computed over ImageNet's huge batches;
# letting them retrain on our smaller batch size can destabilize the whole model.
for layer in base_model.layers[FINE_TUNE_AT:]:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),  # ~100x lower than Phase 1's 1e-3
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

callbacks_phase2 = [
    tf.keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=4, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-7),
    tf.keras.callbacks.ModelCheckpoint("phase2_best.keras", monitor="val_accuracy", save_best_only=True),
]

history_phase2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,           # lower ceiling than Phase 1 — fine-tuning converges (and overfits) faster
    class_weight=class_weights,
    callbacks=callbacks_phase2,
)
```
